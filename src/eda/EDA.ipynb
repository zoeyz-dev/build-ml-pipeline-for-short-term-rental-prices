# EDA - NYC Airbnb

Load sample.csv from W&B and explore the data before building the pipeline.

```python
import wandb
import pandas as pd

run = wandb.init(project="nyc_airbnb", group="eda", save_code=True)
local_path = wandb.use_artifact("sample.csv:latest").file()
df = pd.read_csv(local_path)
df.head()
```

```python
from ydata_profiling import ProfileReport

profile = ProfileReport(df, title="NYC Airbnb", minimal=True)
profile
```

Drop price outliers and convert last_review to datetime. Missing values stay for now.

```python
min_price = 10
max_price = 350

df = df.drop_duplicates()
df = df.dropna(subset=["price"])
idx = df["price"].between(min_price, max_price)
df = df[idx].copy()
df["last_review"] = pd.to_datetime(df["last_review"])

df.info()
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
df["price"].hist(bins=50, ax=axes[0])
axes[0].set_title("Price")
df.groupby("neighbourhood_group")["price"].median().plot(kind="bar", ax=axes[1])
axes[1].set_title("Median price by borough")
plt.tight_layout()
plt.show()
```
